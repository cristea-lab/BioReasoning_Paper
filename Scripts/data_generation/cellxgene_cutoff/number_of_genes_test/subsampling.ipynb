# Subsample 1,000 Cells from an h5ad Dataset
This notebook randomly subsamples 1,000 cells from an h5ad file using a specified random seed and saves the result.

```python
import numpy as np
import scanpy as sc
```

```python
input_file = '../../../../Outputs/cellxgene_cutoff_random_10k_v2/h5ad/subsampled_10k_random.h5ad'  # Replace with your input file path
output_file = '../../../../Outputs/cellxgene_cutoff_random_1k_number_of_genes_test_r1_0528/h5ad/subsampled_1k.h5ad'  # Replace with your desired output file path
n_cells = 1000  # Number of cells to sample
seed = 100  # Random seed for reproducibility
```

```python
adata = sc.read_h5ad(input_file)
print(f"Loaded AnnData with {adata.n_obs} cells and {adata.n_vars} genes.")
```

```python
np.random.seed(seed)
sampled_names = np.random.choice(adata.obs_names, size=n_cells, replace=False)
adata_sub = adata[sampled_names, :].copy()
print(f"Subsampled {adata_sub.n_obs} cells.")
```

```python
adata_sub.write_h5ad(output_file)
print(f"Saved subsampled dataset to {output_file}.")
```
