```python
# Imports
import scanpy as sc
import numpy as np
```

```python
INPUT_FILE  = "/cristealab/xiwang/CellXGene/normal_after_20230515/merged_h5ad/merged_normal_filtered.h5ad"   # <-- put your .h5ad path here
OUTPUT_FILE = "../../../..//Outputs/cellxgene_cutoff_random_10k_v2/h5ad/subsampled_10k_random.h5ad"     # <-- desired output path
N_CELLS     = 10000               # number of cells to keep
RANDOM_SEED = 42                   # set to any integer for reproducibility
```

```python
# 1) Load the AnnData object
adata = sc.read_h5ad(INPUT_FILE)
print(f"Loaded {adata.n_obs:,} cells × {adata.n_vars:,} genes")
```

```python
# 2) Randomly choose N_CELLS indices with the given seed
if N_CELLS > adata.n_obs:
    raise ValueError(f"Cannot subsample {N_CELLS:,} cells from "
                     f"a dataset with only {adata.n_obs:,} cells!")

rng = np.random.default_rng(RANDOM_SEED)
sample_idx = rng.choice(adata.n_obs, size=N_CELLS, replace=False)
```

```python
# 3) Slice the AnnData object (obs_idx, var_idx)
adata_subset = adata[sample_idx, :].copy()
print(f"Subset shape → {adata_subset.shape}")
```

```python
print(f"Subset shape → {adata.shape}")
```

```python
adata_subset.obs_names_make_unique()
```

```python
# 4) Save the subset
adata_subset.write_h5ad(OUTPUT_FILE)
print(f"Wrote subset to {OUTPUT_FILE}")
```
